## Task 1: Finding the Largest Value

The goal is to find the largest number in a dataset by checking each value and comparing it with the current largest value.

```python
numbers = [12, 45, 7, 89, 23, 56, 34]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Dataset:", numbers)
print("Largest value:", largest)
```

### Time Complexity Analysis

The algorithm checks each number in the dataset to find the largest value.

- **Best Case:** O(n)
- **Average Case:** O(n)
- **Worst Case:** O(n)

The time complexity is O(n), also called linear time, because the number of comparisons increases as the dataset becomes larger.

The space complexity is O(1) because the algorithm uses only a fixed number of additional variables to find the largest value.

## Understanding Big O Notation

Big O notation describes how an algorithm's performance changes as the input size increases.

Common time complexities:

- **O(1) — Constant Time:** The execution time stays approximately the same regardless of input size.
- **O(log n) — Logarithmic Time:** The problem size is reduced with each step, as in binary search.
- **O(n) — Linear Time:** The algorithm processes each item in the input.
- **O(n log n) — Linearithmic Time:** Common in efficient sorting algorithms.
- **O(n²) — Quadratic Time:** The work can grow with the square of the input size, as in simple nested-loop sorting algorithms.

## Task 2: Linear Search

Linear search checks each element sequentially until the target value is found. If the target is not in the list, the algorithm checks every element before reporting that it was not found.

```python
def linear_search(numbers, target):
    for index in range(len(numbers)):
        if numbers[index] == target:
            return index
    return -1


numbers = [10, 25, 30, 45, 50, 65]
target = 45

result = linear_search(numbers, target)

print("Dataset:", numbers)
print("Target:", target)

if result != -1:
    print("Target found at index:", result)
else:
    print("Target not found.")
```

### Linear Search Complexity Analysis

- **Best Case:** O(1), when the target is the first element.
- **Average Case:** O(n), because the algorithm may check many elements.
- **Worst Case:** O(n), when the target is the last element or is not present.
- **Space Complexity:** O(1), because the algorithm uses a fixed amount of additional space.

## Task 3: Binary Search

Binary search finds a target by repeatedly checking the middle element of a sorted list. If the target is smaller or larger than the middle element, the search continues in the appropriate half.

```python
def binary_search(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left <= right:
        middle = (left + right) // 2

        if numbers[middle] == target:
            return middle
        elif numbers[middle] < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1


numbers = [10, 20, 30, 40, 50, 60, 70, 80]
target = 60

result = binary_search(numbers, target)

print("Sorted dataset:", numbers)
print("Target:", target)

if result != -1:
    print("Target found at index:", result)
else:
    print("Target not found.")
```

### Binary Search Complexity Analysis

- **Best Case:** O(1), when the middle element is the target.
- **Average Case:** O(log n), because the search space is repeatedly divided in half.
- **Worst Case:** O(log n), when several steps are needed to find the target or determine that it is absent.
- **Space Complexity:** O(1), because the iterative algorithm uses a fixed number of additional variables.

## Task 4: Bubble Sort

Bubble sort repeatedly compares adjacent elements and swaps them if they are in the wrong order. The process continues until the list is sorted.

```python
def bubble_sort(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        for j in range(0, n - i - 1):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

    return numbers


numbers = [64, 34, 25, 12, 22, 11, 90]

print("Original list:", numbers)
print("Sorted list:", bubble_sort(numbers))
```

### Bubble Sort Complexity Analysis

- **Best Case:** O(n²) for this implementation.
- **Average Case:** O(n²).
- **Worst Case:** O(n²).
- **Space Complexity:** O(n), because the function creates a copy of the input list.

Bubble sort is simple to understand, but it can be inefficient for large datasets because it repeatedly compares adjacent elements.

## Task 5: Selection Sort

Selection sort repeatedly finds the smallest element in the unsorted portion of a list and swaps it with the first element of that portion. This continues until the list is sorted.

```python
def selection_sort(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if numbers[j] < numbers[min_index]:
                min_index = j

        numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

    return numbers


numbers = [64, 25, 12, 22, 11]

print("Original list:", numbers)
print("Sorted list:", selection_sort(numbers))
```

### Selection Sort Complexity Analysis

- **Best Case:** O(n²).
- **Average Case:** O(n²).
- **Worst Case:** O(n²).
- **Space Complexity:** O(n), because the function creates a copy of the input list.

Selection sort performs repeated comparisons to find the smallest element in the unsorted portion. It uses a simple approach but can be inefficient for large datasets.

## Task 6: Insertion Sort

Insertion sort builds the sorted portion of a list one element at a time. Each new element is compared with the elements before it and inserted into its correct position.

```python
def insertion_sort(numbers):
    numbers = numbers.copy()

    for i in range(1, len(numbers)):
        key = numbers[i]
        j = i - 1

        while j >= 0 and numbers[j] > key:
            numbers[j + 1] = numbers[j]
            j -= 1

        numbers[j + 1] = key

    return numbers


numbers = [12, 11, 13, 5, 6]

print("Original list:", numbers)
print("Sorted list:", insertion_sort(numbers))
```

### Insertion Sort Complexity Analysis

- **Best Case:** O(n), when the list is already sorted.
- **Average Case:** O(n²).
- **Worst Case:** O(n²), when the list is in reverse order.
- **Space Complexity:** O(n), because the function creates a copy of the input list.

Insertion sort can be efficient for small or nearly sorted datasets, but it can become slow when sorting large, unsorted lists.

## Task 7: Comparing Sorting Algorithms

This task compares the execution times of Bubble Sort, Selection Sort, and Insertion Sort using datasets of different sizes. The results help show how the performance of each algorithm changes as the dataset grows.

```python
import random
import time

sizes = [100, 300, 500]
random.seed(42)

print(f"{'Size':>6} {'Bubble (s)':>14} {'Selection (s)':>16} {'Insertion (s)':>15}")

for size in sizes:
    data = [random.randint(0, 10000) for _ in range(size)]

    timings = {}

    algorithms = [
        ("Bubble", bubble_sort),
        ("Selection", selection_sort),
        ("Insertion", insertion_sort)
    ]

    for name, algorithm in algorithms:
        start = time.perf_counter()
        sorted_data = algorithm(data)
        elapsed = time.perf_counter() - start

        timings[name] = elapsed

        assert sorted_data == sorted(data)

    print(
        f"{size:>6} "
        f"{timings['Bubble']:>14.6f} "
        f"{timings['Selection']:>16.6f} "
        f"{timings['Insertion']:>15.6f}"
    )
```

### Analysis and Interpretation

The execution times show how the performance of Bubble Sort, Selection Sort, and Insertion Sort changes as the dataset size increases. All three algorithms have an average and worst-case time complexity of O(n²), meaning their execution time generally increases as the amount of data grows. The actual results may vary depending on the computer and the input data. This activity demonstrates why understanding time complexity is important when choosing an algorithm.
